```python
import pandas as pd
import matplotlib.pyplot as plt
# %conda install seaborn
import seaborn as sns

url = "https://raw.githubusercontent.com/nkmwicz/data-for-students/refs/heads/main/soccer-players.csv"
df = pd.read_csv(url)
df.head
```

```python
# anatomy of a viz
plt.figure(figsize=(5,3))
plt.xlabel("Height in CM")
plt.ylabel("Frequency")
plt.title("Histograms of soccer player heights(cm)")
plt.xlim(0,210)
plt.ylim(0,500)
plt.grid(True)
text = "skew=0.12"
plt.text(x=0.05, 
         y=0.95, 
         s=text, 
         ha="center", 
         va="center", 
         fontsize=12, 
         color="hotpink",
         transform=plt.gca().transAxes)
plt.axvline(x=105, color="r", linestyle="--", label="Mean")
plt.axhline(y=250, color="b", linestyle="--", label="Center")
plt.legend()
plt.show()
```

```python
import scipy.stats as stats
```

```python
# anatomy of a viz
# Histogram in PLT
plt.figure(figsize=(5,3))
hts=df["height_in_cm"]
plt.hist(x=hts, bins=12, edgecolor= "black", color= "hotpink")
plt.xlabel("Height in CM")
plt.ylabel("Frequency")
plt.title("Histograms of soccer player heights(cm)")
plt.grid(True)
sk = stats.skew (hts, nan_policy="omit")
kurt= stats.kurtosis (hts, nan_policy= "omit")
text = f"Skew: {sk:.3f}\nKurtosis:{kurt:.3f}"
plt.text(x=0.05, 
         y=0.95, 
         s=text, 
         ha="center", 
         va="center", 
         fontsize=12, 
         color="hotpink",
         transform=plt.gca().transAxes)
plt.axvline(x=hts.mean(), color="purple", linestyle="--", label="Mean")
plt.axvline(x=hts.median(), color="green", linestyle="--", label="Median")
plt.legend()
plt.show()
```

```python
# Histogram in SNS
df1=df.loc[df["height_in_cm"]>150]
plt.figure(figsize=(5,3))
sns.set_theme(style = "white")
hist = sns.histplot(data=df1, x="height_in_cm", kde=True, line_kws={"color": "red"})
plt.xlabel("height(cm)")
hts=df["height_in_cm"]
sk = stats.skew (hts, nan_policy="omit")
kurt= stats.kurtosis (hts, nan_policy= "omit")
text = f"Skew: {sk:.3f}\nKurtosis:{kurt:.3f}"
plt.text(x=0.05, 
         y=0.95, 
         s=text, 
         ha="center", 
         va="center", 
         fontsize=12, 
         color="hotpink",
         transform=plt.gca().transAxes)
plt.axvline(x=hts.mean(), color="red", linestyle="--", label="Mean")
plt.axvline(x=hts.median(), color="green", linestyle="--", label="Median")
plt.legend()
plt.show()
```
