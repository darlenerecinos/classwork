```python
import pandas as pd

df = pd.read_csv("./data/french-ambs.csv")
df.head()
```

```python
for col in df.columns:
    print(f"{col} is numeric?: {pd.api.types.is_numeric_dtype(df[col])}")
```

```python
from random import choice, seed

seed(41)

def my_function(x):
    print(x)
df["date_start"] = df["year"].apply(lambda x:f"{x}-{choice(range(1,13)):02d}-{choice(range(1,29))}")

df.head()
```

```python
# drop the column year_start
df.drop(columns=["date_start"])
        
```

```python
# making the year_start better by using a named function
import numpy as np
def get_year_end(row):
    rows = df.loc[
        (df["name"]==row["name"])&
        (df["place"]==row["place"])&
        (df["year"]==row["year"]+1)
    ]
    if len(rows) > 0:
        return np.nan
    date_string = f"{row["year"]}-{choice(range(1,13)):02d}-{choice(range(1,29)):02d}"
    return date_string

df["date_end"]= df.apply(get_year_end, axis=1) # on a df, must pass axis
df.head()

    
```

```python
# bfill: back fill ffill: forward fill
df = df.sort_values(["place","name","year"],ascending=True).reset_index()
df.head()
```

```python
df["date_start"] =  df["date_start"].ffill()
df["date_end"] = df["date_end"].bfill()
df.head()
```

```python
import datetime
fmt = "%Y-%M-%d"

df["date_start"] =  df["date_start"].apply(datetime.datetime.strptime())
```

```python
df.groupby(["place"]).count()
```

```python
df.groupby(["place"]).agg(
    {
        "place": "count",
        "year": "min",
    }
)
```

```python
df.groupby(["place"]).agg(
    place_years=("place", "count"),
    max=("year","max"),
    min=("year", "min"),
    num_ambs=("name","nunique"),
    ambs=("name", "unique")    
).reset_index(drop=True)
```
